```python
#Using the text from the verdict, we will tokenize it into words and punctuation marks.

with open ("the-verdict.txt", "r", encoding = "utf-8") as f:
    raw_text = f.read()

print("Total number of characters:", len(raw_text))

print(raw_text[:99])
```

```python
#Necessary import to strip whitespace and punctuation from the tokens
import re

#splitting the text into tokens using regex to capture punctuation and whitespace as separate tokens
preprocessed = re.split(r'([,.;:?_!"()\']|--|\s)', raw_text)

#Removing empty strings and whitespace-only tokens from the list
preprocessed = [item.strip() for item in preprocessed if item.strip()]

print(preprocessed[:25])

print(len(preprocessed))
```

Now, convert tokens to token_id

```python
all_words = sorted(set(preprocessed))

print(len(all_words))
```

```python
#Create vocab: sort all tokens in alphabetical order, then assign a unique int to each token
vocab = {token: integer for integer, token in enumerate(all_words)}

print(len(vocab))
```

```python
class SimpleTokenizer:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {v: k for k, v in vocab.items()}

    def encode(self, text):

        preprocessed = re.split(r'([,.;:?_!"()\']|--|\s)', text)

        preprocessed = [item.strip() for item in preprocessed if item.strip()]

        ids = [self.str_to_int[token] for token in preprocessed if token in self.str_to_int]

        return ids

    def decode(self, ids):

        text = " ".join([self.int_to_str[token_id] for token_id in ids if token_id in self.int_to_str])

        text = re.sub(r'\s+([,.;:?_!"()\']|--)', r'\1', text)

        return text
```

```python
tokenizer = SimpleTokenizer(vocab)

text = """It's the last he painted, you know," Mrs. Gisburn said with pardonable pride."""

ids = tokenizer.encode(text)

print(ids)

print(tokenizer.decode(ids))
```

```python
class SimpleTokenizerV2:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {v: k for k, v in vocab.items()}

    def encode(self, text):

        preprocessed = re.split(r'([,.;:?_!"()\']|--|\s)', text)

        preprocessed = [item.strip() for item in preprocessed if item.strip()]

        ids = [self.str_to_int.get(token, self.str_to_int["<|unk|>"]) for token in preprocessed]

        return ids

    def decode(self, ids):
        text = " ".join([self.int_to_str.get(token_id, "<|unk|>") for token_id in ids])

        text = re.sub(r'\s+([,.;:?_!"()\']|--)', r'\1', text)

        return text
```
