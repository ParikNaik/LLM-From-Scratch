```python
import torch.nn as nn
```

```python
class SelfAttentionV2(nn.Module):

    def __init__ (self, dim_in, dim_out, qkv_bias = False):
    
        super().__init__()
        self.W_query = nn.Linear(nn.rand(dim_in,dim_out), bias = qkv_bias)
        self.W_key = nn.Linear(nn.rand(dim_in,dim_out), bias = qkv_bias)
        self.W_value = nn.Linear(nn.rand(dim_in,dim_out), bias = qkv_bias)

    def Forward(self, x):
        queries = self.W_query(x)
        keys = self.W_keys(x)
        values = self.W_values(x)

        attn_scores = queries @ keys.T
        attention_scores = attn_scores/ (keys.shape[-1] ** 0.5)
        attn_weights = nn.Softmax(attention_scores, dim = -1)

        context_vect = attn_weights @ values
        return context_vect
```
